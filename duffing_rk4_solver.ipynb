```python
import numpy
```

```python
damping_ratio = 0.05
natural_frequency = 1.0
cubic_stiffness = 1.0

initial_position = 0.5
initial_velocity = 0.0

start_time = 0.0
end_time = 50.0
step_size = 0.001
```

```python
def state_derivative(state, damping_ratio, natural_frequency, cubic_stiffness):
    position, velocity = state
    acceleration = (2 * damping_ratio * natural_frequency * velocity
                    + natural_frequency**2 * position
                    - cubic_stiffness * position**3)
    return numpy.array([velocity, acceleration])
```

```python
def runge_kutta_4_step(state, step_size, damping_ratio, natural_frequency, cubic_stiffness):
    equation_parameters = (damping_ratio, natural_frequency, cubic_stiffness)
    first_slope = state_derivative(state, *equation_parameters)
    second_slope = state_derivative(state + step_size / 2 * first_slope, *equation_parameters)
    third_slope = state_derivative(state + step_size / 2 * second_slope, *equation_parameters)
    fourth_slope = state_derivative(state + step_size * third_slope, *equation_parameters)
    return state + step_size / 6 * (first_slope + 2 * second_slope + 2 * third_slope + fourth_slope)
```

```python
def solve_with_runge_kutta_4(initial_position, initial_velocity, start_time, end_time, step_size,
                             damping_ratio, natural_frequency, cubic_stiffness):
    number_of_steps = int(numpy.ceil((end_time - start_time) / step_size))
    times = numpy.linspace(start_time, end_time, number_of_steps + 1)
    states = numpy.empty((number_of_steps + 1, 2))
    states[0] = initial_position, initial_velocity
    for step_index in range(number_of_steps):
        states[step_index + 1] = runge_kutta_4_step(states[step_index], times[step_index + 1] - times[step_index],
                                                    damping_ratio, natural_frequency, cubic_stiffness)
    return times, states[:, 0], states[:, 1]
```

```python
times, positions, velocities = solve_with_runge_kutta_4(initial_position, initial_velocity, start_time, end_time,
                                                        step_size, damping_ratio, natural_frequency, cubic_stiffness)
energies = 0.5 * velocities**2 - 0.5 * natural_frequency**2 * positions**2 + 0.25 * cubic_stiffness * positions**4
```

```python
print(f'{"time":>10} {"position":>22} {"velocity":>22} {"energy":>22}')
for sample_index in numpy.linspace(0, len(times) - 1, 11).astype(int):
    print(f'{times[sample_index]:10.3f} {positions[sample_index]:22.15f} '
          f'{velocities[sample_index]:22.15f} {energies[sample_index]:22.15f}')
```
